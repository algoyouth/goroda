```python
import pandas as pd
import random

df = pd.read_excel('03_all_cities.xlsx', sheet_name='cities')

# Список городов
goroda = df['city']
```

```python
# Множество букв, с которых начинаются города
start_letters = {city[0] for city in goroda}

# Множество букв, которыми оканчиваются города
end_letters = {city[-1] for city in goroda}
```

```python
all_letters = set('АБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯ')
start_dict = {}
for letter in all_letters:
    start_dict[letter] = set()
for city in goroda:
    l = city[0].upper()
    if l in start_dict:
        start_dict[l].add(city)
r = random.choice(city)
print(r)
```

```python
g = random.choice(all_letters)
print("Начинаем на букву ", g)
while g != ".":
    gg = random.choice(list(start_dict[g[-1].upper()]))
    print(gg)
    g1 = gg[-1].upper()
    if g1 not in start_dict:
        g1 = gg[-2].upper()
    g = input("Введите город на букву ", g1)
    if g[-1].upper() != g1:
        print("Вы ввели город на другую букву. Вы проиграли")
        break
    
```

```python


# Словарь, где ключи - буквы, с которых начинаются города, а значения - количество слов
start_count_dict = {letter: sum(1 for city in goroda if city.startswith(letter)) for letter in start_letters}

# Словарь, где ключи - буквы, с которых начинаются города, а значения - списки городов
start_city_dict = {letter: [city for city in goroda if city.startswith(letter)] for letter in start_letters}

# Множество букв, на которые не начинаются названия городов
all_letters = set('АБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯ')
not_starting_letters = all_letters - start_letters

# Множество букв, на которые не оканчивается ни один город
not_ending_letters = all_letters - end_letters

# Вывод результатов
print("Начинающиеся буквы:", start_letters)
print("Заканчивающиеся буквы:", end_letters)
print("Словарь количества городов по начальным буквам:", start_count_dict)
print("Словарь городов по начальным буквам:", start_city_dict)
print("Буквы, на которые не начинаются названия городов:", not_starting_letters)
print("Буквы, на которые не оканчивается ни один город:", not_ending_letters)
```
